```python
from pyspark.sql.functions import *
```

```python
silver_patients = spark.table("workspace.silver.patients")
silver_encounters = spark.table("workspace.silver.encounters")
silver_conditions = spark.table("workspace.silver.conditions")
silver_observations = spark.table("workspace.silver.observations")
silver_medications = spark.table("workspace.silver.medications")
silver_claim = spark.table("workspace.silver.claims")
```

```python
# Total number of patients
total_patients = silver_patients.count()

# Number of records with null patient_id
null_patient_id_count = silver_patients.filter(col("patient_id").isNull()).count()

# Check for duplicate patient_id records
duplicate_patient_ids = silver_patients.groupBy("patient_id").count().filter(col("count") > 1)
duplicate_count = duplicate_patient_ids.count()

# Number of records with null DOB
null_dob_count = silver_patients.filter(col("DOB").isNull()).count()

# Number of records with invalid DOB (e.g., future dates)
invalid_dob_count = silver_patients.filter(col("DOB") > current_date()).count()

# Number of records with null or invalid gender
null_gender_count = silver_patients.filter(col("gender").isNull()).count()
invalid_gender_count = silver_patients.filter(~col("gender").isin("male", "female", "other")).count()

# Display results
display(spark.createDataFrame([
    ("Total Patients", total_patients),
    ("Null patient_id", null_patient_id_count),
    ("Duplicate patient_id", duplicate_count),
    ("Null DOB", null_dob_count),
    ("Invalid DOB", invalid_dob_count),
    ("Null gender", null_gender_count),
    ("Invalid gender", invalid_gender_count)
], ["Metric", "Count"]))
```
