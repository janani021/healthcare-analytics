```python
# Dimension tables
dim_patient = spark.table("gold.dim_patient")
dim_condition = spark.table("gold.dim_condition")
dim_date = spark.table("gold.dim_date")

# Fact tables
fact_encounter = spark.table("gold.fact_encounter")
fact_patient_conditions = spark.table("gold.fact_patient_conditions")

# Semantic/dashboard tables
gold_population_health = spark.table("gold.gold_population_health")
gold_clinical_analytics = spark.table("gold.gold_clinical_analytics")
```

```python
from pyspark.sql.functions import * 

# Assert no nulls in patient_key
assert dim_patient.filter(col("patient_key").isNull()).count() == 0, "Nulls found in patient_key"

# Assert no nulls in gender
assert dim_patient.filter(col("gender").isNull()).count() == 0, "Nulls found in gender"

# Assert row count is greater than zero
assert dim_patient.count() > 0, "dim_patient table is empty"

# Check for duplicates in patient_key
dup_count = dim_patient.groupBy("patient_key").count().filter(col("count") > 1).count()
assert dup_count == 0, "Duplicates found in patient_key"
```

```python
# Assert no nulls in condition_key
assert dim_condition.filter(col("condition_key").isNull()).count() == 0, "Nulls found in condition_key"

# Assert no nulls in condition_code
assert dim_condition.filter(col("condition_code").isNull()).count() == 0, "Nulls found in condition_code"

# Assert row count is greater than zero
assert dim_condition.count() > 0, "dim_condition table is empty"

# Check for duplicates in condition_key
dup_count_condition = dim_condition.groupBy("condition_key").count().filter(col("count") > 1).count()
assert dup_count_condition == 0, "Duplicates found in condition_key"
```

```python
# Assert no nulls in date_key
assert dim_date.filter(col("date_key").isNull()).count() == 0, "Nulls found in date_key"

# Assert row count is greater than zero
assert dim_date.count() > 0, "dim_date table is empty"

# Check for duplicates in date_key
dup_count_date = dim_date.groupBy("date_key").count().filter(col("count") > 1).count()
assert dup_count_date == 0, "Duplicates found in date_key"
```

```python
# Assert no nulls in encounter_key
assert fact_encounter.filter(col("encounter_key").isNull()).count() == 0, "Nulls found in encounter_key"

# Assert no nulls in patient_key
assert fact_encounter.filter(col("patient_key").isNull()).count() == 0, "Nulls found in patient_key"

# Assert row count is greater than zero
assert fact_encounter.count() > 0, "fact_encounter table is empty"

# Check for duplicates in encounter_key
dup_count_encounter = fact_encounter.groupBy("encounter_key").count().filter(col("count") > 1).count()
assert dup_count_encounter == 0, "Duplicates found in encounter_key"
```

```python
# Assert no nulls in patient_key
assert fact_patient_conditions.filter(col("patient_key").isNull()).count() == 0, "Nulls found in patient_key"

# Assert no nulls in condition_key
assert fact_patient_conditions.filter(col("condition_key").isNull()).count() == 0, "Nulls found in condition_key"

# Assert no nulls in date_key
assert fact_patient_conditions.filter(col("date_key").isNull()).count() == 0, "Nulls found in date_key"

# Assert row count is greater than zero
assert fact_patient_conditions.count() > 0, "fact_patient_conditions table is empty"

# Check for duplicates in patient_key, condition_key, and date_key
dup_count_fpc = fact_patient_conditions.groupBy("patient_key", "condition_key", "date_key").count().filter(col("count") > 1).count()
assert dup_count_fpc == 0, "Duplicates found in (patient_key, condition_key, date_key)"
```
